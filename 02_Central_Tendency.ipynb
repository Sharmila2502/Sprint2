# 1. Mean (Arithmetic Mean)

## Definition

The mean is the sum of all the values in a dataset divided by the
number of values.

It is commonly called the average.

The mean gives us a single value that represents the overall
central value of the dataset.

## Formula

Population mean:

μ = Σx / N

Sample mean:

x̄ = Σx / n

Where:
- Σx = Sum of all observations
- N = Number of observations in the population
- n = Number of observations in the sample
- μ = Population mean
- x̄ = Sample mean

## Important Point

The mean uses every value in the dataset. Because of this, extreme
values (outliers) can significantly affect the mean.

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

mean_value = np.mean(marks)

print("Marks:", marks)
print("Mean:", mean_value)
```

# 2. Median

## Definition

The median is the middle value of a dataset after arranging
the values in ascending or descending order.

It divides the ordered dataset into two halves.

- Approximately 50% of the observations are at or below the median.
- Approximately 50% of the observations are at or above the median.

## Formula

For an odd number of observations:

Median = Value at position (n + 1) / 2

For an even number of observations:

Median = Average of the values at positions n / 2 and (n / 2) + 1

Here, n is the number of observations.

## Important Point

The median is less affected by extreme values than the mean.

The data must be sorted before identifying the middle value.

```python
import numpy as np

# Odd number of observations
data_odd = np.array([10, 20, 30, 40, 50])

# Even number of observations
data_even = np.array([10, 20, 30, 40])

median_odd = np.median(data_odd)
median_even = np.median(data_even)

print("Odd dataset median:", median_odd)
print("Even dataset median:", median_even)
```

# 3. Mode

## Definition

The mode is the value or category that occurs most frequently
in a dataset.

A dataset can have:

- One mode: Unimodal
- Two modes: Bimodal
- More than two modes: Multimodal
- No mode under a strict frequency definition, if all values
  occur equally often

## Formula / Method

There is no arithmetic formula like the mean.

Mode = Value with the highest frequency

Frequency means the number of times a value occurs.

## Important Point

Mode can be used for both numerical and categorical data.

For categorical data, such as product type or payment method,
the mode identifies the most common category.

```python
from collections import Counter

data = [10, 20, 20, 30, 40]

frequency = Counter(data)

highest_frequency = max(frequency.values())

modes = [
    value
    for value, count in frequency.items()
    if count == highest_frequency
]

print("Frequency:", frequency)
print("Mode:", modes)
print("Highest Frequency:", highest_frequency)
```

# 4. Weighted Mean

## Definition

The weighted mean is an average in which different values
contribute differently according to their assigned weights.

A value with a larger weight has a greater influence on the
final average.

The ordinary mean treats every observation equally. The
weighted mean does not.

## Formula

Weighted Mean = Σ(wx) / Σw

Where:

- x = Individual value
- w = Weight assigned to that value
- wx = Value multiplied by its weight
- Σ(wx) = Sum of weighted values
- Σw = Sum of all weights

## Important Point

Weights represent relative importance or contribution.

If all weights are equal, the weighted mean becomes the
ordinary arithmetic mean.

```python
import numpy as np

marks = np.array([80, 90, 70])
weights = np.array([3, 2, 1])

weighted_mean = np.average(marks, weights=weights)

print("Marks:", marks)
print("Weights:", weights)
print("Weighted Mean:", weighted_mean)
```

# 5. Geometric Mean

## Definition

The geometric mean is an average calculated by multiplying
positive values together and taking the nth root, where n
is the number of values.

It is particularly useful for quantities that combine
multiplicatively, such as growth factors and investment returns.

## Formula

For n positive values:

GM = (x1 × x2 × ... × xn)^(1/n)

Equivalent formula:

GM = (Πxi)^(1/n)

Where:

- xi = Each observation
- n = Number of observations
- Π = Product of all observations

## Important Point

The standard geometric mean requires positive values.

For growth rates, use growth factors rather than directly
averaging the percentages.

For example, a growth rate of 10% corresponds to a growth
factor of 1.10.

```python
import numpy as np

data = np.array([2, 8])

geometric_mean = np.exp(np.mean(np.log(data)))

print("Data:", data)
print("Geometric Mean:", geometric_mean)
```

# 6. Harmonic Mean

## Definition

The harmonic mean is a type of average calculated by taking
the reciprocal of the arithmetic mean of the reciprocals
of the observations.

It is useful for averaging rates and ratios when the
observations share an appropriate common basis.

## Formula

HM = n / Σ(1/xi)

Where:

- n = Number of observations
- xi = Each observation
- 1/xi = Reciprocal of each observation

## Important Point

For the standard harmonic mean, the observations must be
positive and nonzero.

The harmonic mean is often appropriate for averaging rates
when the same amount of work, distance, or another suitable
quantity is involved in each case.

The correct type of average depends on the situation.

```python
import numpy as np

speeds = np.array([30, 60])

harmonic_mean = len(speeds) / np.sum(1 / speeds)

print("Speeds:", speeds)
print("Harmonic Mean:", harmonic_mean)
```
