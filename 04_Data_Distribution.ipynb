**Data Distribution**

**Normal Distribution**

A normal distribution is a continuous, bell-shaped distribution in which most observations are near the mean, and observations become less common as we move farther away from the mean.

Its main characteristics are:

It is symmetric around the mean.

The mean, median, and mode are equal.

The two tails extend in both directions.

Its shape is determined by its mean and standard deviation.

The mean controls the center, while the standard deviation controls how spread out the distribution is.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate 1000 normally distributed test scores
scores = np.random.normal(loc=70, scale=10, size=1000)

print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Standard deviation:", np.std(scores))

plt.hist(scores, bins=30, density=True)
plt.title("Normal Distribution of Test Scores")
plt.xlabel("Test Score")
plt.ylabel("Density")
plt.show()
```

**Gaussian Distribution**

A Gaussian distribution is another name for a normal distribution. The terms are commonly used interchangeably in statistics.

A Gaussian distribution is described by:

μ — the mean, which determines its center.

σ — the standard deviation, which determines its spread.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)

# Simulate bottle filling measurements
volumes = np.random.normal(loc=500, scale=5, size=1000)

plt.hist(volumes, bins=30, density=True, alpha=0.6)

# Display a Gaussian-shaped curve using the sample mean and std
mean = np.mean(volumes)
std = np.std(volumes)

x = np.linspace(min(volumes), max(volumes), 200)

pdf = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

plt.plot(x, pdf)
plt.title("Gaussian Distribution of Bottle Volumes")
plt.xlabel("Volume (ml)")
plt.ylabel("Density")
plt.show()
```

**Uniform Distribution**

A uniform distribution is a distribution in which values within a specified interval have equal probability density for a continuous variable.

For a continuous uniform distribution between a and b:

The distribution is flat across the interval.

Values outside the interval have zero probability density.

The mean is (a+b)/2.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate 1000 values between 0 and 10
values = np.random.uniform(low=0, high=10, size=1000)

print("Mean:", np.mean(values))
print("Minimum:", np.min(values))
print("Maximum:", np.max(values))

plt.hist(values, bins=20, density=True, edgecolor="black")
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

**Symmetric Distribution**

A symmetric distribution has a shape where the left side is a mirror image of the right side around its center.

For a symmetric distribution:

The two sides have matching shapes.

The mean and median are equal when the mean exists.

A symmetric distribution does not have to be normal or bell-shaped.

A uniform distribution is one example of a symmetric distribution. A normal distribution is another.

```python
import numpy as np
import matplotlib.pyplot as plt

data = [2, 4, 6, 8, 10]

print("Mean:", np.mean(data))
print("Median:", np.median(data))

plt.hist(data, bins=5, edgecolor="black")
plt.title("Example of Symmetric Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

**Left Skew (Negative Skew)**

A left-skewed distribution, also called a negatively skewed distribution, has a longer tail extending toward smaller values.

Its typical pattern is:

Mean<Median

The mean is often pulled toward the long left tail by unusually low observations. This is a common pattern, not a universal rule for every possible dataset described as left-skewed

```python
import numpy as np
import matplotlib.pyplot as plt

# Most values are high, with a smaller number of low values
data = [1, 7, 8, 8, 9, 9, 9, 10, 10, 10, 10]

print("Mean:", np.mean(data))
print("Median:", np.median(data))

plt.hist(data, bins=10, edgecolor="black")
plt.title("Left-Skewed Data Example")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

**Right Skew (Positive Skew)**

A right-skewed distribution, also called a positively skewed distribution, has a longer tail extending toward larger values.

Its typical pattern is:

Mean>Median

A few unusually large values can pull the mean toward the right tail.

```python
import numpy as np
import matplotlib.pyplot as plt

# Many small values and a few large values
spending = [10, 12, 15, 18, 20, 22, 25, 30, 35, 200]

print("Mean:", np.mean(spending))
print("Median:", np.median(spending))

plt.hist(spending, bins=10, edgecolor="black")
plt.title("Right-Skewed Customer Spending")
plt.xlabel("Spending")
plt.ylabel("Frequency")
plt.show()
```

**Kurtosis**

Kurtosis is a statistical measure related to the heaviness of a distribution's tails compared with a normal distribution. It is also influenced by extreme observations.

A common convention is to use excess kurtosis, where a normal distribution has a value of 0.

Mesokurtic: excess kurtosis is approximately 0; normal-like tail reference.

Leptokurtic: positive excess kurtosis; heavier tails relative to the normal reference.

Platykurtic: negative excess kurtosis; lighter tails relative to the normal reference.

Kurtosis should not be interpreted as simply measuring how tall or flat the center looks.

```python
import numpy as np
from scipy.stats import kurtosis

data = [10, 12, 13, 14, 15, 16, 17, 18, 20, 100]

print("Excess kurtosis:", kurtosis(data))
```

**Empirical Rule (68–95–99.7)**

The Empirical Rule describes how data are distributed around the mean when the distribution is approximately normal.

For a normal distribution:

About 68% of observations lie within 1 standard deviation of the mean.

About 95% lie within 2 standard deviations.

About 99.7% lie within 3 standard deviations.

```python
import numpy as np

mean = 70
std = 10

# Generate normally distributed scores
np.random.seed(42)
scores = np.random.normal(mean, std, 10000)

within_1_std = np.mean((scores >= mean - std) & (scores <= mean + std))
within_2_std = np.mean((scores >= mean - 2 * std) & (scores <= mean + 2 * std))
within_3_std = np.mean((scores >= mean - 3 * std) & (scores <= mean + 3 * std))

print("Within 1 SD:", within_1_std * 100, "%")
print("Within 2 SD:", within_2_std * 100, "%")
print("Within 3 SD:", within_3_std * 100, "%")
```
