**Probability**

**Introduction to Probability:**

Probability is a number that describes how likely an event is to happen.

Probability ranges from 0 to 1:

0 means the event is impossible.

1 means the event is certain.

A value between 0 and 1 represents varying degrees of likelihood.

When all outcomes are equally likely:

P(A)=  Total number of possible outcomes / Number of outcomes favorable to A

Here, P(A) means the probability of event A.

```python
favorable_outcomes = 1
total_outcomes = 6

probability = favorable_outcomes / total_outcomes

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

**Events:**

An event is a collection of one or more outcomes from a random experiment.

For a die roll:

Event A: rolling an even number = {2, 4, 6}

Event B: rolling a number greater than 4 = {5, 6}

Event C: rolling a 7 = {} because it is impossible with a standard die.

An event can contain one outcome or several outcomes.

```python
sample_space = {1, 2, 3, 4, 5, 6}

# Event A: rolling an even number
event_A = {2, 4, 6}

probability_A = len(event_A) / len(sample_space)

print("Event A:", event_A)
print("Probability of A:", probability_A)
```

**Sample Space:**

The sample space is the set of all possible outcomes of a random experiment. It is often represented by S or Ω.

Examples:

Tossing a coin: {Heads, Tails}

Rolling a die: {1, 2, 3, 4, 5, 6}

Tossing two coins: {HH, HT, TH, TT}

A sample space must include every possible outcome relevant to the experiment being modeled.

```python
import itertools

coin = ["H", "T"]

# Generate all possible outcomes for two coin tosses
sample_space = list(itertools.product(coin, repeat=2))

print("Sample space:", sample_space)
print("Number of outcomes:", len(sample_space))

# Exactly one head
event_A = [outcome for outcome in sample_space if outcome.count("H") == 1]

print("Exactly one head:", event_A)
print("Probability:", len(event_A) / len(sample_space))
```

**Independent Events:**

Two events are independent if the occurrence of one does not change the probability of the other.

For independent events A and B:

P(A∩B)=P(A)P(B)

The symbol A∩B means that both A and B occur.

```python
p_A = 0.5
p_B = 0.5

# Probability that both independent events occur
p_both = p_A * p_B

print("Probability of both heads:", p_both)
```

**Dependent Events:**

Two events are dependent when the occurrence of one changes the probability of the other.

For example, drawing a card from a deck and not replacing it changes the cards available for the next draw.

For dependent events:

P(A∩B)=P(A)P(B∣A)

Here, P(B∣A) is the probability of B occurring given that A has occurred.

```python
p_first_red = 3 / 5

# After drawing a red ball, 2 red balls remain out of 4
p_second_red_given_first_red = 2 / 4

p_both_red = p_first_red * p_second_red_given_first_red

print("Probability of drawing two red balls:", p_both_red)
```

**Conditional Probability:**

Conditional probability is the probability of an event occurring when we know that another event has occurred.


```python
python_students = 18
both_python_and_sql = 10

p_sql_given_python = both_python_and_sql / python_students

print("P(SQL | Python):", p_sql_given_python)
print("Percentage:", p_sql_given_python * 100, "%")
```

**Bayes' Theorem:**

Bayes' Theorem helps us update the probability of an event when new evidence becomes available.

It is especially useful when we know the probability of seeing evidence under different conditions, and want to reason backward from the evidence

```python
p_fraud = 0.01
p_flag_given_fraud = 0.90
p_flag_given_legitimate = 0.05

p_legitimate = 1 - p_fraud

# Total probability of receiving a flag
p_flag = (
    p_flag_given_fraud * p_fraud
    + p_flag_given_legitimate * p_legitimate
)

# Bayes' theorem
p_fraud_given_flag = (
    p_flag_given_fraud * p_fraud
) / p_flag

print("Probability of being flagged:", p_flag)
print("Probability of fraud given flag:", p_fraud_given_flag)
print("Percentage:", p_fraud_given_flag * 100, "%")
```

**Probability Rules:**

Probability rules help calculate the probability of combined events.

1. Addition Rule (The "Or" Rule) Used to find the probability that either event \[A\] or event \[B\] occurs
2. Multiplication Rule (The "And" Rule) Used to find the probability that both event \[A\] and event \[B\] occur together
3. Complement Rule (The "Not" Rule)Used to find the probability that an event will not happen


```python
p_A = 0.4
p_B = 0.3
p_A_and_B = 0.1

# Addition rule
p_A_or_B = p_A + p_B - p_A_and_B

# Complement rule
p_neither = 1 - p_A_or_B

print("P(A or B):", p_A_or_B)
print("P(neither A nor B):", p_neither)
```

**Random Variables:**

A random variable assigns a numerical value to each possible outcome of a random experiment.

Two common types are:

Discrete random variable: takes countable values, such as the number of heads in three coin tosses.

Continuous random variable: can take values across a continuous range, such as waiting time or a person's height.

A random variable is often represented by a capital letter such as X.

```python
import itertools
from collections import Counter

coin = ["H", "T"]

outcomes = list(itertools.product(coin, repeat=2))

# X = number of heads
head_counts = [outcome.count("H") for outcome in outcomes]

frequencies = Counter(head_counts)
total = len(outcomes)

for value in sorted(frequencies):
    probability = frequencies[value] / total
    print(f"P(X = {value}) = {probability}")
```

**Expected Value:**

The expected value is the long-run average value of a random variable over repeated trials under the same probability model.

```python
outcomes = [100, 0]
probabilities = [0.2, 0.8]

expected_value = sum(
    value * probability
    for value, probability in zip(outcomes, probabilities)
)

print("Expected payout: ₹", expected_value)
```
