**Probability Distributions**

**Bernoulli Distribution:**

A Bernoulli distribution models a single experiment that has only two possible outcomes:

Success, represented by 1

Failure, represented by 0

The probability of success is p, and the probability of failure is 1−p

```python
P(X=x)=px(1−p)1−x,x∈{0,1}

Where:

p = probability of success

1−p = probability of failure

x can be 0 or 1

Its expected value is E[X]=p, and its variance is p(1−p)
```

Where it is used

A customer either purchases a product or does not.

An email is classified as spam or not spam.

A machine component either fails or does not fail during a specified period.

A user either clicks an advertisement or does not click it.

**AI/ML example:**

In a binary classification task, a model might estimate the probability that a transaction is fraudulent.

For example:

Fraud = 1

Not fraud = 0

If the estimated probability of fraud is 0.8, a decision threshold might be used to convert that probability into a predicted class. The threshold depends on the task and the costs of different errors.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, poisson, uniform, norm, expon
p = 0.3

x = [0, 1]
probabilities = [1 - p, p]

plt.bar(x, probabilities)
plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.title("Bernoulli Distribution (p = 0.3)")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.ylim(0, 1)
plt.show()
```

**Binomial Distribution**


A Binomial distribution models the number of successes in a fixed number of independent trials, where each trial has two possible outcomes and the same probability of success.

A random variable X follows a binomial distribution if:

X∼Binomial(n,p)

The assumptions are:

The number of trials n is fixed.

Each trial has two possible outcomes.

Trials are independent.

The probability of success p is constant across trials.

**Where it is used:**

Counting how many customers purchase out of a fixed group.

Counting successful quality checks in a batch.

Counting heads in a fixed number of coin tosses.

Estimating the number of successful deliveries in a fixed set of attempts, when the assumptions are reasonable.

**AI/ML example:**

A business may model the number of users who click an advertisement out of a fixed number of users, assuming each user's click outcome is independent and has the same click probability.

This can help estimate conversion counts and quantify uncertainty in experiments. In practice, user behavior may not be independent or have identical probabilities, so the assumptions should be checked.

```python
n = 10
p = 0.2

x = np.arange(0, n + 1)
probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)
plt.title("Binomial Distribution (n = 10, p = 0.2)")
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.xticks(x)
plt.show()
```

**Poisson Distribution:**

A Poisson distribution models the number of times an event occurs in a fixed interval of time, space, or another defined exposure, when events occur at a constant average rate and independently under the Poisson model assumptions.

A random variable X follows a Poisson distribution if:

X∼Poisson(λ)

Here, λ is the expected number of events in the specified interval.

**Where it is used:**

Number of calls received by a support center per minute.

Number of system errors in an hour.

Number of customer arrivals in a given time period.

Number of defects per unit, when the model assumptions are appropriate

**AI/ML example:**

Poisson models can be used for count prediction, such as predicting the number of support tickets a server may generate during a time window.

Poisson regression is one model used for count-valued targets. If the data are overdispersed or violate the model assumptions, alternatives such as negative binomial regression may be more suitable.

```python
lam = 4

x = np.arange(0, 16)
probabilities = poisson.pmf(x, mu=lam)

plt.bar(x, probabilities)
plt.title("Poisson Distribution (lambda = 4)")
plt.xlabel("Number of events")
plt.ylabel("Probability")
plt.xticks(x)
plt.show()
```

**Uniform Distribution:**

A continuous uniform distribution models a value that can fall anywhere within a fixed interval, with constant probability density throughout that interval.

It is written as:

X∼U(a,b)

Where a is the lower bound and b is the upper bound.

**Where it is used:**

Simulating a random time within a fixed interval.

Random sampling within a bounded range.

Modeling a quantity when a constant density is a reasonable simplifying assumption.

Generating random values for experiments.

**AI/ML example:**

Uniform distributions can be used to randomly sample hyperparameter values within a range. For example, a simulation might sample a bounded parameter uniformly when that sampling scheme is appropriate.

For parameters spanning several orders of magnitude, log-uniform sampling may be more suitable than ordinary uniform sampling.

```python
a = 0
b = 10

x = np.linspace(a - 2, b + 2, 500)
density = uniform.pdf(x, loc=a, scale=b - a)

plt.plot(x, density)
plt.title("Continuous Uniform Distribution (0 to 10)")
plt.xlabel("Value")
plt.ylabel("Probability density")
plt.show()
```

Statistical inference involving normally distributed data or errors.

Modeling noise in some scientific and engineering applications.

Not every real-world variable follows a normal distribution.

**AI/ML example:**

Normal distributions are used in Gaussian Naive Bayes, some generative models, and models of continuous noise. They can also help describe residual errors when normality is a reasonable assumption.

```python
mu = 500
sigma = 5

x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 500)
density = norm.pdf(x, loc=mu, scale=sigma)

plt.plot(x, density)
plt.title("Normal Distribution (mean = 500, SD = 5)")
plt.xlabel("Volume (ml)")
plt.ylabel("Probability density")
plt.show()
```

**Exponential Distribution:**

An exponential distribution is a continuous distribution often used to model the waiting time between events in a Poisson process.

It is written as:

X∼Exponential(λ)

Here, λ>0 is the event rate per unit of time. A larger rate means a shorter average waiting time.

**Where it is used:**

Modeling time between arrivals in a Poisson process.

Modeling some component lifetimes under a constant hazard-rate assumption.

Estimating waiting times between events.

Queueing and reliability models.

**AI/ML example:**

Exponential distributions can be used in survival and reliability analysis when a constant hazard rate is reasonable. They can also model inter-arrival times in simulations of event-driven systems.

```python
lam = 6  # events per hour

# SciPy's scale parameter is the reciprocal of the rate
scale = 1 / lam

x = np.linspace(0, 1.5, 500)  # time in hours
density = expon.pdf(x, scale=scale)

plt.plot(x, density)
plt.title("Exponential Distribution (rate = 6 per hour)")
plt.xlabel("Waiting time (hours)")
plt.ylabel("Probability density")
plt.show()
```
