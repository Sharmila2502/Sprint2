**Statistical_Tests**

**T-Test:**

A t-test is used to determine whether the mean of one group differs significantly from another mean.

It is commonly used when:

The data is numerical.
We are comparing means.
The population standard deviation is unknown.
The sample size may be relatively small.
Types of t-test

There are three common versions:

One-sample t-test
Compare one sample mean against a known/reference value.
Independent two-sample t-test
Compare means of two independent groups.
Paired t-test
Compare measurements from the same subjects/items before and after something.
Example

Suppose:

Average sales before training = ₹50,000
Average sales after training = ₹55,000

We can test whether the increase is statistically significant.

**Assumptions:**

For the usual t-test:

Data is numerical.
Observations are independent for an independent t-test.
Data is approximately normally distributed, especially for small samples.
For the standard independent t-test, the groups have equal variances.

**Business Example:**

An e-commerce company wants to know:

Did a new website design increase average order value?

Two groups:

Old website customers
New website customers

We compare their average order values.

```python
import numpy as np
from scipy.stats import ttest_ind

old_website = np.array([450, 470, 490, 500, 510, 530])
new_website = np.array([500, 520, 540, 550, 570, 590])

t_stat, p_value = ttest_ind(
    old_website,
    new_website,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

**Z-Test:**

A Z-test is used to determine whether a sample mean or proportion differs significantly from a reference value.

It is closely related to the t-test.

The important difference is that a traditional Z-test assumes that the population standard deviation is known, whereas a t-test generally estimates it from the sample.

Suppose a company says:

Average delivery time is 30 minutes.

You collect a sample of deliveries and get:

Sample mean = 32 minutes

You want to know:

Is 32 significantly different from 30?

A Z-test can be used when its assumptions are satisfied.

**Assumptions:**

Observations are independent.
Data is approximately normally distributed, or sample size is sufficiently large for the sampling distribution to be approximately normal.
Population standard deviation is known for a traditional one-sample Z-test.
Appropriate measurement scale and sampling design.

**Business Example:**

A food-delivery company historically has:
Average delivery time = 30 minutes
Population standard deviation = 5 minutes

A sample of 100 deliveries has:
Sample mean = 31.2 minutes

We want to determine whether the current average differs from 30 minutes

```python
import numpy as np
from statsmodels.stats.weightstats import ztest

delivery_times = np.array([
    28, 30, 31, 32, 29,
    33, 30, 31, 32, 30
])

z_stat, p_value = ztest(
    delivery_times,
    value=30
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

**Chi-Square Test:**

The Chi-Square test is mainly used with categorical data.

There are two important forms:

1. Chi-Square Test of Independence

Tests whether two categorical variables are associated.

Example:  Gender × Product Purchased

2. Chi-Square Goodness-of-Fit

Tests whether observed category frequencies match an expected distribution

Observed customers:
Product A = 20
Product B = 30
Product C = 50

**Definition:**

The Chi-Square test determines whether differences between observed categorical frequencies and expected frequencies are statistically significant.

**Assumptions:**

Data is categorical/count data.
Observations are independent.
Categories are mutually exclusive.
Expected frequencies should generally be sufficiently large.

A common practical rule is that expected counts should not be too small; if they are, alternatives such as Fisher's exact test may be appropriate for certain contingency tables.

**When to use:**

Use Chi-Square when:

Working with categorical variables.
Testing association between categorical variables.
Testing observed versus expected frequencies.

```python
import pandas as pd
from scipy.stats import chi2_contingency

data = pd.DataFrame(
    [
        [40, 30, 20],
        [20, 35, 30]
    ],
    columns=["Electronics", "Clothing", "Grocery"],
    index=["Member", "Non-Member"]
)

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:")
print(expected)
```

**ANOVA**

**Definition:**

ANOVA = Analysis of Variance

ANOVA is used to determine whether the means of three or more groups differ significantly.

Instead of performing many t-tests, we can use one ANOVA test.

**Assumptions:**

Observations are independent.
Dependent variable is numerical.
Data within groups is approximately normally distributed.
Variances are approximately equal across groups for standard one-way ANOVA.

**When to use**

Use ANOVA when:

Comparing means of 3 or more groups.
Dependent variable is numerical.
Groups are independent for one-way independent ANOVA.

```python
from scipy.stats import f_oneway

north = [50, 52, 48, 51, 53]
south = [60, 62, 59, 61, 63]
west = [55, 57, 54, 56, 58]

f_stat, p_value = f_oneway(
    north,
    south,
    west
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

**Mann-Whitney U Test:**

Definition

The Mann-Whitney U test is a non-parametric test used to compare two independent groups.

It is often used as an alternative to the independent two-sample t-test when the assumptions for the t-test are not appropriate.

Instead of directly comparing means, it works with the ranks of observations.

Example

Suppose customer spending is:

Group A: 100, 110, 120, 130
Group B: 200, 210, 220, 230

The distribution of spending between the groups appears very different.

Mann-Whitney U can test whether the two groups differ in their distributions/ranks.

**Assumptions:**

Two independent groups.
Observations are independent.
Outcome is ordinal or continuous.
If interpreting the test specifically as a difference in distributions/location, the shapes of the distributions should be considered.

```python
from scipy.stats import mannwhitneyu

website_a = [2, 3, 3, 4, 4, 5]
website_b = [3, 4, 4, 5, 5, 5]

u_stat, p_value = mannwhitneyu(
    website_a,
    website_b,
    alternative="two-sided"
)

print("U-statistic:", u_stat)
print("P-value:", p_value)
```

**Shapiro-Wilk Test:**

**Definition:**

The Shapiro-Wilk test is used to test whether a sample is consistent with a normal distribution.

This is especially useful when checking assumptions before applying tests such as:

t-test
ANOVA

Hypotheses

H₀  :  The data comes from a normal distribution.

H₁  :  The data does not come from a normal distribution.

This is important because the interpretation is the opposite direction from many tests.

Assumptions:


Observations should be independent.
Data should be numerical/continuous for the usual use case.
The test is intended to assess normality of the sample.

**Business Example:**

Suppose we want to use a t-test to compare employee productivity scores.

Before applying the t-test, we can check whether the scores are approximately normally distributed.

```python
from scipy.stats import shapiro

productivity = [
    72, 75, 78, 80, 82,
    79, 77, 81, 76, 83
]

stat, p_value = shapiro(productivity)

print("Shapiro-Wilk statistic:", stat)
print("P-value:", p_value)
```
