Data cleaning means finding and correcting problems in the dataset before giving the data to a machine-learning model.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [25, 30, 28, -5, 200],
    "Salary": [30000, 40000, 35000, 50000, 60000]
})

print(df.describe())
```

Before training a model, we need to make sure the input data is meaningful.

**Missing Value Handling:**

Statistics helps us decide what value should replace the missing value.

**Mean Imputation:**

Use the mean when the data is reasonably symmetric and does not contain extreme outliers

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

**Median Imputation:**

Median is often better when the data is skewed or contains outliers

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

**Mode Imputation:**

For categorical variables:

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

**AI/ML Use:**

Many ML algorithms cannot directly handle missing values, so statistics helps us choose an appropriate replacement.

**Outlier Detection**

An outlier is an observation that is unusually far from the rest of the data.

10, 12, 11, 13, 12, 15, 100

100 is likely an outlier.

Statistics provides several methods

**Method 1: IQR**

First calculate:

   IQR = Q3 - Q1

then,

  Lower Bound = Q1 - 1.5 × IQR
  Upper Bound = Q3 + 1.5 × IQR

Values outside these limits can be flagged as potential outliers.

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower) |
    (df["Salary"] > upper)
]

print(outliers)
```

**Method 2: Z-score**

Z-score tells us how many standard deviations an observation is from the mean.

  Z = (X - Mean) / Standard Deviation

  |Z| > 3 → possible outlier

**AI/ML Use:**

Outliers can strongly affect models such as:
