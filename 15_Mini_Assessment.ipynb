**Explain the difference between Mean and Median.**

Mean is the average value of a dataset.

Example:

10, 20, 30

Mean = (10 + 20 + 30) / 3 = 20

Median is the middle value when data is arranged in ascending or descending order.

Example:

10, 20, 30, 40, 50

Median = 30

If there are even number of values, take the average of the two middle values.

**What is Variance?**

Variance measures how far the values are spread from the mean.

A high variance means the values are widely spread.

A low variance means the values are close to the mean.

**What is Standard Deviation?**

Standard deviation measures the spread of data around the mean.

SD=sqrt(Variance)

If standard deviation is small, the data points are close to the mean.

**Explain Bayes' Theorem with an example.**

Bayes' Theorem calculates the probability of an event based on previously known information.

Example:

Suppose:

1% of people have a disease.

A test is positive for 90% of people who have the disease.

The test is also positive for 5% of healthy people.

Bayes' theorem can be used to calculate:  Probability that a person actually has the disease given that the test is positive.

This is used in Naive Bayes classification, spam detection, medical diagnosis, and fraud detection.

**Difference between Population and Sample**

| Population             | Sample                            |
| ---------------------- | --------------------------------- |
| Entire group           | Subset of the population          |
| Usually larger         | Usually smaller                   |
| Example: All customers | Example: 1,000 selected customers |


Example:

If a company has 1 million customers, all 1 million are the population. Selecting 10,000 customers for analysis creates a sample.

**Explain Correlation and Covariance.**

**Covariance**

Covariance tells us the direction of the relationship between two variables.

Positive covariance → variables tend to increase together.
Negative covariance → one tends to increase when the other decreases.

**Correlation**

Correlation measures the strength and direction of a linear relationship.

Its value ranges from:  −1 to +1

+1 → perfect positive relationship

-1 → perfect negative relationship

0 → no linear relationship

Covariance tells mainly the direction; correlation tells direction and standardized strength.

**What is Skewness?**

Skewness measures the asymmetry of a distribution.

**Positive Skew**

The distribution has a longer tail toward the right.

Most values → lower
Few very high values → right tail

**Negative Skew**

The distribution has a longer tail toward the left.

Few very low values → left tail
Most values → higher


**Explain IQR.**

IQR stands for Interquartile Range.

It measures the spread of the middle 50% of the data.

    IQR=Q3−Q1

Where:

Q1 = 25th percentile

Q3 = 75th percentile

**Outlier detection**

  Lower Bound = Q1 - 1.5(IQR) 

  Upper Bound = Q3 + 1.5(IQR)

Values outside these bounds are considered potential outliers.

**What is a p-value?**

A p-value is the probability of obtaining results at least as extreme as the observed results, assuming the null hypothesis is true.

Common interpretation using α = 0.05:

p < 0.05 → reject the null hypothesis.

p ≥ 0.05 → do not reject the null hypothesis.

A p-value does not tell us the probability that the null hypothesis itself is true.

**Difference between Type I and Type II Error**

| Type I Error                                      | Type II Error                                                               |
| ------------------------------------------------- | --------------------------------------------------------------------------- |
| Rejecting a true null hypothesis                  | Failing to reject a false null hypothesis                                   |
| False Positive                                    | False Negative                                                              |
| Example: Saying a treatment works when it doesn't | Saying there is insufficient evidence of an effect when one actually exists |


**Python code to calculate Variance**

```python
import numpy as np

data = [10, 20, 30, 40, 50]

variance = np.var(data)

print("Variance:", variance)
```

```python
variance = np.var(data, ddof=1)

print("Sample Variance:", variance)
```

**Detect Outliers using the IQR Method**

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 15, 14, 13, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_bound) | (data > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

**Calculate Correlation between Two Variables**

```python
import numpy as np

X = [10, 20, 30, 40, 50]
Y = [15, 25, 35, 45, 55]

correlation = np.corrcoef(X, Y)[0, 1]

print("Correlation:", correlation)
```
