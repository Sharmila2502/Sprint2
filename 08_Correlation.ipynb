**Correlation & Covariance**

**Correlation:**

Correlation measures the strength and direction of the relationship between two variables.

For example, we might investigate whether the number of hours a student studies is related to their exam marks.

Correlation helps answer questions such as:

Do two variables tend to increase together?

Does one tend to increase when the other decreases?

Is there a clear relationship between them?

For Pearson correlation, the coefficient r ranges from −1 to +1.

```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80]
})

correlation = data["Study_Hours"].corr(data["Marks"])

print("Pearson correlation:", correlation)
```

**Covariance:**

Covariance measures whether two variables tend to move together.

Positive covariance: the variables tend to move in the same direction.

Negative covariance: the variables tend to move in opposite directions.

Covariance near zero: there is little linear co-movement indicated by covariance.

```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3],
    "Marks": [40, 50, 60]
})

# Sample covariance matrix
covariance_matrix = data.cov()

print(covariance_matrix)
```

**Positive Correlation:**

Positive correlation means that higher values of one variable tend to be associated with higher values of another variable.

Examples:

Study hours and marks may have a positive correlation.

Advertising expenditure and sales may have a positive correlation.

Temperature and ice cream sales may have a positive correlation in some settings.

A positive correlation does not mean that every observation follows the trend exactly.

**AI/ML example:**

A data scientist may find that a house's floor area and its price are positively correlated in a particular housing dataset.

This relationship can help with exploratory analysis, but other features—such as location, age, and condition—may also affect price.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Marks": [35, 45, 55, 65, 75, 85]
})

r = data["Study_Hours"].corr(data["Marks"])
print("Correlation:", r)

plt.scatter(data["Study_Hours"], data["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Positive Correlation")
plt.show()
```

**Negative Correlation:**

Negative correlation means that higher values of one variable tend to be associated with lower values of another variable.

Examples:

Vehicle age and resale value may be negatively correlated.

Product price and quantity demanded may be negatively correlated under certain market conditions.

Distance remaining and fuel consumed during a journey may show a negative relationship in a suitable dataset.

**AI/ML example:**

A data scientist studying used cars may find that vehicle age is negatively correlated with resale price.

This can be useful during feature exploration for a price-prediction model. The strength of the relationship may vary by car type, mileage, and condition

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Car_Age": [1, 2, 3, 4, 5, 6],
    "Price": [900000, 800000, 700000, 600000, 500000, 400000]
})

r = data["Car_Age"].corr(data["Price"])
print("Correlation:", r)

plt.scatter(data["Car_Age"], data["Price"])
plt.xlabel("Car Age (years)")
plt.ylabel("Price")
plt.title("Negative Correlation")
plt.show()
```

**No Correlation:**

No correlation usually means that a correlation coefficient indicates little or no linear relationship between two variables.

However, a coefficient close to zero does not rule out a nonlinear relationship.


**Business and AI/ML example:**

A retailer might investigate whether a customer's shoe size is linearly related to their spending. The variables may show little or no linear association in a particular dataset.

A data scientist should still consider whether nonlinear patterns, subgroups, or other variables are relevant.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "X": [-3, -2, -1, 0, 1, 2, 3],
    "Y": [9, 4, 1, 0, 1, 4, 9]
})

r = data["X"].corr(data["Y"])
print("Pearson correlation:", r)

plt.scatter(data["X"], data["Y"])
plt.xlabel("X")
plt.ylabel("Y")
plt.title("A Nonlinear Relationship with Zero Pearson Correlation")
plt.show()
```

**Pearson Correlation:**

Pearson correlation measures the strength and direction of the linear relationship between two numerical variables.

Pearson correlation:

Ranges from −1 to +1.

Is sensitive to outliers.

Measures linear association, not general nonlinear association.

Is undefined if either variable has zero variance.

```python
import pandas as pd
from scipy.stats import pearsonr

study_hours = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 70, 80]

r, p_value = pearsonr(study_hours, marks)

print("Pearson correlation:", r)
print("P-value:", p_value)
```

**Spearman Correlation:**

Spearman correlation measures the strength and direction of a monotonic relationship using the ranks of the observations rather than their original values.

A monotonic relationship generally moves in one direction: as one variable increases, the other tends to increase or tends to decrease, though the pattern need not be a straight line.

Spearman correlation is useful when:

The relationship is monotonic but not linear.

Data are ordinal, such as rankings or rating scales.

You want a rank-based measure that is less sensitive to the exact numerical distances between values.

It can still be affected by outliers that change the ordering of observations.

```python
from scipy.stats import spearmanr

study_hours = [1, 2, 3, 4, 5]
marks = [40, 45, 60, 80, 95]

rho, p_value = spearmanr(study_hours, marks)

print("Spearman correlation:", rho)
print("P-value:", p_value)
```

**Correlation Matrix:**

A correlation matrix displays the pairwise correlation coefficients among several variables.

For example, with three features—study hours, attendance, and marks—the matrix contains the correlations for every pair.

Properties of a Pearson correlation matrix:

The diagonal entries are 1 when each variable has nonzero variance, because each variable is perfectly correlated with itself.

The matrix is symmetric.

Each coefficient lies between −1 and +1.

Pairwise correlations describe associations between two variables at a time; they do not automatically account for other variables.

**AI/ML example:**

In a regression problem, a data scientist may examine a correlation matrix to identify features that are strongly linearly associated with each other.

Strong correlations between predictors can indicate possible multicollinearity in some models, such as ordinary least squares regression. It is a signal to investigate further—not an automatic instruction to remove a feature

```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Attendance": [50, 60, 70, 80, 90],
    "Marks": [40, 50, 65, 75, 85]
})

correlation_matrix = data.corr(method="pearson")

print(correlation_matrix.round(2))
```

**Heatmap:**

A heatmap is a visual representation of numerical values using colors.

When applied to a correlation matrix, each cell represents the correlation between a pair of variables. The color makes it easier to spot positive, negative, and near-zero correlations across many features.

A heatmap is a visualization tool; it does not calculate correlation by itself. We calculate the matrix first, then plot it.

**Business and AI/ML example:**

A business analyst might use a correlation heatmap to explore relationships among sales, advertising, discounts, customer visits, and revenue.

An AI/ML practitioner might use one during exploratory data analysis to inspect feature relationships and investigate possible redundancy. The heatmap alone cannot establish causation or determine which features should be removed.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Attendance": [50, 60, 65, 75, 85, 95],
    "Marks": [35, 45, 55, 68, 78, 90],
    "Sleep_Hours": [8, 7, 7, 6, 6, 5]
})

# Calculate the correlation matrix
corr_matrix = data.corr()

# Draw the heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
