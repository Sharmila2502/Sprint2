# 1. Range

## Definition

Range is a measure of dispersion that tells us the difference
between the largest and smallest values in a dataset.

It gives a simple indication of how widely the observations
are spread from the minimum to the maximum.

## Formula

Range = Maximum Value - Minimum Value

## Important Points

- A larger range indicates a wider span of observed values.
- A smaller range indicates a narrower span of observed values.
- Range depends only on the minimum and maximum values.
- Range is sensitive to extreme values and outliers.

```python
import numpy as np

sales = np.array([10, 15, 20, 25, 30])

data_range = np.max(sales) - np.min(sales)

print("Sales:", sales)
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))
print("Range:", data_range)
```

# 2. Variance

## Definition

Variance measures how far the observations are spread around
their mean.

It is calculated by finding the difference between each
observation and the mean, squaring those differences, and
then averaging them.

Squaring prevents positive and negative deviations from
cancelling each other out.

## Formula

Population variance:

σ² = Σ(xi - μ)² / N

Sample variance:

s² = Σ(xi - x̄)² / (n - 1)

Where:

- xi = Each observation
- μ = Population mean
- x̄ = Sample mean
- N = Population size
- n = Sample size

## Important Points

- Population variance divides by N.
- Sample variance commonly divides by n - 1.
- Variance is always non-negative.
- A variance of zero means all observations are identical.
- Variance is expressed in squared units.

```python
import numpy as np

data = np.array([2, 4, 6])

population_variance = np.var(data)
sample_variance = np.var(data, ddof=1)

print("Population Variance:", population_variance)
print("Sample Variance:", sample_variance)
```

# 3. Standard Deviation

## Definition

Standard deviation measures the spread of observations around
their mean.

It is the square root of variance.

Unlike variance, standard deviation is expressed in the same
units as the original data, which often makes it easier to
interpret.

## Formula

Population standard deviation:

σ = √[Σ(xi - μ)² / N]

Sample standard deviation:

s = √[Σ(xi - x̄)² / (n - 1)]

## Important Points

- A low standard deviation means observations are relatively
  close to the mean.
- A high standard deviation means observations are more spread
  out around the mean.
- Standard deviation is zero when all observations are equal.
- Standard deviation is sensitive to extreme values.

```python
import numpy as np

data = np.array([2, 4, 6])

population_std = np.std(data)
sample_std = np.std(data, ddof=1)

print("Population Standard Deviation:", population_std)
print("Sample Standard Deviation:", sample_std)
```

# 4. Quartiles

## Definition

Quartiles divide an ordered dataset into four parts.

They describe the values at the 25%, 50%, and 75% positions
of the distribution.

The three main quartiles are:

- Q1: First quartile, or 25th percentile
- Q2: Second quartile, or 50th percentile (the median)
- Q3: Third quartile, or 75th percentile

## Formula / Method

Quartiles are positional statistics.

First, arrange the observations in ascending order.

Then determine the 25th, 50th, and 75th percentile values
using the selected quartile convention.

Different statistical software packages may use different
interpolation conventions, so quartile values can sometimes
differ slightly.

## Important Point

Quartiles help us understand the distribution of data and
are used to calculate the interquartile range.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

# 5. Percentiles

## Definition

A percentile indicates a value below which a specified
percentage of observations in a distribution fall.

For example, the 25th percentile is a value associated with
the lower 25% of the distribution.

The 50th percentile is the median.

The 90th percentile is a value associated with the lower
90% of the distribution.

## Formula / Method

Percentiles are calculated from ordered data using a chosen
percentile convention.

A common positional formula for a dataset of n ordered
observations is:

Position = (p / 100) × (n + 1)

Where:

- p = Desired percentile
- n = Number of observations

This is one convention; other methods use different
positions or interpolation rules.

## Important Point

A percentile is a position in a distribution, not the
percentage of the value itself.

```python
import numpy as np

scores = np.array([40, 50, 60, 70, 80, 90, 100])

p25 = np.percentile(scores, 25)
p50 = np.percentile(scores, 50)
p90 = np.percentile(scores, 90)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("90th Percentile:", p90)
```

# 6. Interquartile Range (IQR)

## Definition

The interquartile range measures the spread of the middle
50% of a dataset.

It is the difference between the third quartile and the
first quartile.

## Formula

IQR = Q3 - Q1

Where:

- Q1 = 25th percentile
- Q3 = 75th percentile

## Outlier Detection Rule

A commonly used rule identifies potential outliers as values:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound are
flagged as potential outliers.

## Important Points

- IQR is less sensitive to extreme values than range.
- It measures the spread of the middle half of the data.
- The 1.5 × IQR rule flags potential outliers; it does not
  prove that those observations are errors.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:", outliers)
```

# 7. Coefficient of Variation (CV)

## Definition

The coefficient of variation measures the size of the
standard deviation relative to the mean.

It expresses variability as a proportion of the mean,
often reported as a percentage.

This makes it useful for comparing relative variability
between datasets with different scales or units.

## Formula

Population CV:

CV = (σ / μ) × 100%

Sample CV:

CV = (s / x̄) × 100%

Where:

- σ = Population standard deviation
- μ = Population mean
- s = Sample standard deviation
- x̄ = Sample mean

## Important Points

- A larger CV indicates greater variability relative to
  the mean.
- A smaller CV indicates less variability relative to
  the mean.
- CV is most meaningful for ratio-scale data with a
  meaningful, positive mean.
- CV is undefined when the mean is zero and can be
  misleading when the mean is near zero.
- Comparing CVs for data with negative means generally
  requires special care.

```python
import numpy as np

machine_a = np.array([8, 10, 12])
machine_b = np.array([80, 100, 120])

mean_a = np.mean(machine_a)
std_a = np.std(machine_a)

mean_b = np.mean(machine_b)
std_b = np.std(machine_b)

cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print("Machine A CV:", cv_a, "%")
print("Machine B CV:", cv_b, "%")
```
