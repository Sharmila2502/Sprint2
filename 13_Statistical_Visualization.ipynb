**Statistical Data Visualization:**

**Histogram**

A histogram shows the distribution of a numerical variable by dividing the values into ranges called bins.

For example, if we have employee salaries:

30, 35, 38, 42, 45, 47, 50, 52, 55...

A histogram groups them into intervals:

30–40
40–50
50–60
60–70

and shows how many observations fall into each interval.

**Purpose:**

A histogram helps us understand:

Distribution
Center
Spread
Skewness
Possible outliers
Multiple peaks

**When to use:**

Use a histogram when:

Your variable is numerical.
You want to understand its distribution.
You want to check whether data is approximately normal.
You want to identify skewness or unusual values.

**Advantages:**

Easy to understand.

Shows the overall distribution.

Helps identify skewness.

Useful during Exploratory Data Analysis.

**Limitations:**

Result depends on bin size.

Not suitable for categorical variables.

Exact individual values are not visible.

Different bin widths can produce different impressions.

```python
import matplotlib.pyplot as plt

salary = [25, 28, 30, 32, 35, 36, 38, 40, 42, 45,
          47, 50, 52, 55, 60, 65, 70]

plt.hist(salary, bins=5)

plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")

plt.show()
```

**Bar Chart:**

A bar chart compares values across categories.

Example:

    Department → Number of Employees

HR       → 20
IT       → 50
Finance  → 30
Sales    → 40

**Purpose:**

To compare the size or frequency of different categories.

**When to use:**

Use a bar chart for:

Categorical data

Comparing departments

Product sales

Number of customers by city

Category counts

**Advantages:**

  Very easy to understand.

  Good for comparing categories.

  Values can be displayed directly.
  
  Works well for rankings.

**Limitations:**
      
Not ideal for showing continuous distributions.

Too many categories can make the chart difficult to read.

Does not show the underlying distribution within each category.

```python
import matplotlib.pyplot as plt

departments = ["HR", "IT", "Finance", "Sales"]
employees = [20, 50, 30, 40]

plt.bar(departments, employees)

plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Employees by Department")

plt.show()
```

**Line Chart:**

A line chart connects data points in an ordered sequence, commonly across time.

Month → Revenue
Jan   → 100
Feb   → 120
Mar   → 150
Apr   → 140

**Purpose:**

To show:

Trends

Changes over time

Growth/decline

Patterns

**When to use:**

Use a line chart for:

Monthly sales

Daily website traffic

Stock prices

Temperature over time

Model performance over epoch


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 150, 140, 170]

plt.plot(months, sales, marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
```

**Scatter Plot**

A scatter plot displays individual observations using two numerical variables

X = Hours Studied
Y = Exam Score

Each student becomes one point.

**Purpose:**

To investigate relationships between two numerical variables.

**It can help identify:**

Positive relationships

Negative relationships

No obvious relationship

Clusters

Outliers

Nonlinear patterns

**When to use:**

Use scatter plots when:

Both variables are numerical.

You want to investigate a relationship.

You want to visually inspect correlation.

**Advantages:**

Shows individual observations.

Good for detecting relationships.

Helps identify outliers.

Can reveal nonlinear patterns.

**Limitations:**

Can become difficult to read with very large datasets due to overplotting.

Correlation does not imply causation.

A scatter plot alone doesn't quantify the strength of a relationship.

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [42, 48, 55, 63, 69, 78, 84, 91]

plt.scatter(hours, scores)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Hours Studied vs Exam Score")

plt.show()
```

**Pie Chart:**

A pie chart represents how categories contribute to a whole.

Product A → 40%
Product B → 30%
Product C → 20%
Product D → 10%

Total: 100%

**When to use:**

**Use a pie chart when:**

Categories represent parts of one total.

There are only a small number of categories.

The proportions are more important than precise values.

**Advantages:**

Simple for a small number of categories.

Visually communicates proportions.

Easy to understand for simple datasets.

**Limitations:**

Difficult to compare similar-sized slices.

Becomes confusing with many categories.

Exact values are harder to compare than with bars.

Not appropriate for trends over time



```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [40, 30, 20, 10]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")

plt.show()
```

**Density Plot:**

A density plot, often called a KDE plot, shows a smoothed estimate of the distribution of a numerical variable.

It is similar to a histogram, but instead of bars, it displays a smooth curve.

**Purpose:**

To understand the shape of a numerical distribution.

**It helps identify:**

Center

Spread

Skewness

Multiple peaks

**When to use:**

**Use a density plot when:**

You have continuous numerical data.

You want to compare distributions.

You want a smoother alternative to a histogram.

**Advantages:**

Smooth representation.

Good for comparing distributions.

Shows overall shape clearly.

**Limitations:**

Depends on the bandwidth/smoothing parameter.

Can hide individual observations.

Can be misleading with very small datasets.

Less intuitive for beginners than a histogram.



```python
import seaborn as sns
import matplotlib.pyplot as plt

salary = [
    25, 28, 30, 32, 35, 36,
    38, 40, 42, 45, 47, 50,
    52, 55, 60, 65, 70
]

sns.kdeplot(salary)

plt.xlabel("Salary")
plt.title("Salary Density Plot")

plt.show()
```

**Box Plot**

A box plot summarizes the distribution of numerical data using:

Minimum/ lower whisker

Q1

Median

Q3

Upper whisker

Potential outliers

**Purpose:**

To quickly understand:

Central tendency

Spread

Skewness

Potential outliers

**When to use:**

**Use box plots when:**


You want to identify outliers.

You want to compare distributions between groups.

You want a compact summary of numerical data.

**Advantages:**

Excellent for detecting potential outliers.

Compact.

Easy to compare multiple groups.

Shows median and quartiles.

**Limitations:**

Doesn't show the complete distribution shape.

For small datasets, individual values may be hidden.

A point outside the whiskers is a potential outlier according to the plotting rule; it is not automatically an error.

```python
import seaborn as sns
import matplotlib.pyplot as plt

salary = [25, 28, 30, 32, 35, 36, 38, 40, 42, 45, 47, 100]

sns.boxplot(y=salary)

plt.ylabel("Salary")
plt.title("Salary Box Plot")

plt.show()
```

**Violin Plot:**

A violin plot combines ideas from:

Box plot

Density plot

**Purpose:**

To compare the distribution and spread of numerical variables across groups.

**Advantages:**

Shows distribution shape.

Shows concentration of observations.

Useful for comparing groups.

Contains more distribution information than a simple box plot.

**Limitations:**
    
More difficult for beginners.

KDE smoothing can affect appearance.

Can be difficult to interpret with very small samples.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": [
        "IT", "IT", "IT",
        "HR", "HR", "HR",
        "Finance", "Finance", "Finance"
    ],
    "Salary": [
        50, 55, 60,
        35, 40, 45,
        45, 50, 55
    ]
}

sns.violinplot(
    x="Department",
    y="Salary",
    data=data
)

plt.title("Salary Distribution by Department")

plt.show()
```

**Pair Plot:**

A pair plot displays relationships between multiple numerical variables at once.

**Purpose:**

To perform quick multivariate exploratory data analysis.

**When to use**

Use a pair plot when:

You have a manageable number of numerical features.

You want to investigate relationships between many variables.

You want to identify possible correlations and clusters.

**Advantages:**

Shows many relationships at once.

Useful during EDA.

Helps identify correlations.

Can reveal clusters and unusual observations.

**Limitations:**

Becomes extremely cluttered with many features.

Computationally expensive for very large datasets.

Interpretation can become difficult.





```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.pairplot(df)

plt.show()
```

**Heatmap:**

A heatmap represents values using different visual intensities.

A very common data-science use is displaying a correlation matrix.

**Purpose**

To visualize a matrix of values.

**Common uses include:**

Correlation matrix

Confusion matrix

Missing-value patterns

Feature relationships

**Advantages:**

Compact.

Easy to identify strong/weak values.

Good for correlation matrices.
    
Useful for high-dimensional EDA.

**Limitations:**
    
Too many variables can make it cluttered.

Color interpretation depends on the color scale.

Correlation heatmaps show association, not causation.

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

correlation = df.select_dtypes(
    include="number"
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
