**Hypothesis Testing**

Hypothesis testing is a statistical method used to evaluate a claim about a population using sample data.

For example, a company claims that its new training program increases employee productivity. We collect sample data and use a statistical test to assess whether the data provide evidence against a specified baseline claim.

The usual process is:

State the null and alternative hypotheses.

Choose a significance level.

Select an appropriate statistical test.

Calculate the test statistic and p-value.

Compare the p-value with the significance level.

Report the conclusion in the context of the question.

A statistical test does not prove a hypothesis with certainty. It evaluates how compatible the observed data are with a specified null model and its assumptions.

**Null Hypothesis:**

The null hypothesis, written as H0, is the baseline statement being tested.

It often represents:

  No difference
  No effect
  No association
  A population parameter equal to a specified value
  

**Business and AI/ML example:**

Business: A company tests whether a new website design changes the average purchase amount.

AI/ML: A researcher compares two models and tests whether their population performance difference is zero, using a test suitable for the evaluation design.

```python
# Null hypothesis:
# The population mean delivery time is 30 minutes.

H0_mean = 30

print("Null hypothesis: population mean = ", H0_mean)
```

**Alternative Hypothesis:**

The alternative hypothesis, written as Ha or H1, is the statement that is considered in contrast to the null hypothesis.

It may claim that a population parameter:

Is different from a specified value.

Is greater than a specified value.

Is less than a specified value.

**Business and AI/ML example:**

A company tests whether a new checkout design changes the average time needed to complete a purchase.

Null hypothesis: The population mean completion time is unchanged.

Alternative hypothesis: The population mean completion time is different.

```python
# Two-tailed hypotheses about average delivery time

H0 = "Population mean delivery time = 30 minutes"
H1 = "Population mean delivery time != 30 minutes"

print("H0:", H0)
print("H1:", H1)
```

**Significance Level**

The significance level, represented by α, is the threshold chosen in advance for deciding when a result is statistically significant.

A commonly used value is:      α=0.05

This corresponds to a 5% long-run Type I error rate when the test assumptions hold and the null hypothesis is true.

The significance level is not the probability that the null hypothesis is true.

```python
alpha = 0.05

p_value = 0.03

if p_value <= alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

**p-value**

A p-value is the probability, assuming the null hypothesis and the test's assumptions are true, of obtaining a test statistic at least as extreme as the one observed, in the direction or directions specified by the test.

In simpler words, it measures how unusual the observed result would be under the null model.

A small p-value indicates that the observed result would be relatively unusual under the null model.

A large p-value indicates that the observed result is not especially unusual under that model.

A p-value is not the probability that the null hypothesis is true, nor is it the probability that the result happened “by chance.”

```python
from scipy.stats import ttest_1samp

delivery_times = [28, 29, 31, 27, 30, 26, 29, 28, 27, 30]

# Test whether the population mean differs from 30
test_statistic, p_value = ttest_1samp(
    delivery_times,
    popmean=30,
    alternative="two-sided"
)

alpha = 0.05

print("Test statistic:", test_statistic)
print("P-value:", p_value)

if p_value <= alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

**Confidence Interval**

A confidence interval (CI) is a range of plausible values for a population parameter, calculated from sample data using a method with a specified long-run coverage rate.

For example, a 95% confidence interval procedure is designed so that, over repeated samples under its assumptions, 95% of intervals constructed by that procedure would contain the true parameter.

A 95% confidence interval does not mean there is a 95% probability that the fixed population parameter lies inside this particular interval, under the usual frequentist interpretation.

Example:

Suppose a survey estimates average customer satisfaction as 8.0, with a 95% confidence interval of 7.5 to 8.5.

The interval communicates uncertainty around the estimate. Its width depends on factors such as sample size, variability, and the confidence level.

```python
import numpy as np
from scipy import stats

scores = np.array([7, 8, 9, 6, 8, 9, 7, 8, 10, 8])

n = len(scores)
mean = np.mean(scores)
standard_error = stats.sem(scores)

# 95% t confidence interval
ci = stats.t.interval(
    confidence=0.95,
    df=n - 1,
    loc=mean,
    scale=standard_error
)

print("Sample mean:", mean)
print("95% confidence interval:", ci)
```

**Type I Error**

A Type I error occurs when we reject the null hypothesis even though it is true.

It is often described as a false positive.

If the significance level is 0.05, the test is designed to have a Type I error probability of at most 5% under the null hypothesis, assuming the test is correctly specified and its assumptions hold.

**Business and AI/ML example:**

Business: A company concludes that a new advertisement changes sales when, in reality, there is no true effect.

AI/ML: A statistical evaluation reports a difference between two models when their true performance difference is zero.

**Type II Error**

A Type II error occurs when we fail to reject the null hypothesis even though it is false.

It is often described as a false negative.

The probability of a Type II error is commonly represented by β.


**Business and AI/ML example:**

Business: A company fails to detect a genuine improvement in sales from a new campaign.

AI/ML: An experiment fails to detect a real performance difference between two models.

**One-tailed Test:**

A one-tailed test evaluates evidence for an effect in one specified direction.

There are two forms:

Right-tailed: tests whether a parameter is greater than a reference value.

Left-tailed: tests whether a parameter is less than a reference value.

```python
from scipy.stats import ttest_1samp

output = [102, 105, 101, 107, 103, 104, 106, 102]

# Right-tailed test: is the mean greater than 100?
t_stat, p_value = ttest_1samp(
    output,
    popmean=100,
    alternative="greater"
)

print("T-statistic:", t_stat)
print("One-tailed p-value:", p_value)
```

**Two-tailed Test**

A two-tailed test evaluates evidence that a parameter differs from a reference value in either direction.

Example:
A company wants to test whether average delivery time differs from 30 minutes.

```python
from scipy.stats import ttest_1samp

delivery_times = [28, 29, 31, 27, 30, 26, 29, 28, 27, 30]

# Two-tailed test: is the mean different from 30?
t_stat, p_value = ttest_1samp(
    delivery_times,
    popmean=30,
    alternative="two-sided"
)

print("T-statistic:", t_stat)
print("Two-tailed p-value:", p_value)
```
