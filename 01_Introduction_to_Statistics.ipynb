# 1. What is Statistics?

Statistics is the process of collecting, organizing, analyzing, interpreting, and presenting data.

In simple words, statistics helps us understand a large amount of information by converting raw data into meaningful information.

For example, suppose we have the marks of 100 students.

Instead of looking at all 100 marks individually, statistics allows us to calculate:

- Average marks
- Highest marks
- Lowest marks
- Middle value
- How much the marks vary
- How the marks are distributed

Therefore, statistics helps us understand patterns and make decisions using data.

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

average = np.mean(marks)
minimum = np.min(marks)
maximum = np.max(marks)

print("Marks:", marks)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

# 2. Why Statistics is Important in AI/ML

Machine learning works with data, and statistics provides tools to understand that data.

Statistics helps an AI/ML engineer to:

1. Understand the dataset
2. Find patterns in data
3. Identify unusual values
4. Measure relationships between variables
5. Summarize large datasets
6. Select useful features
7. Evaluate machine-learning models
8. Understand uncertainty in predictions

For example, before training a model, we may need to know:

- What is the average value?
- Are there missing values?
- Are there outliers?
- How are values distributed?
- Are two variables related?

Statistics provides methods to answer these questions.

```python
import numpy as np

customers = np.array([1000])
churned = np.array([200])

churn_rate = (churned / customers) * 100

print("Total Customers:", customers[0])
print("Churned Customers:", churned[0])
print("Churn Rate:", churn_rate[0], "%")
```

20% of customers have churned.

This simple statistical measurement gives us useful information about the dataset before building a churn prediction model.

# 3. Statistics vs Machine Learning

Statistics and Machine Learning both work with data, but their main goals are different.

Statistics traditionally focuses on understanding data, relationships, uncertainty, and drawing conclusions about a population.

Machine Learning focuses mainly on learning patterns from data so that a model can make predictions or decisions on new data.

Statistics asks questions such as:

- What is the average?
- Is there a relationship between variables?
- How confident are we about our result?
- Can we make conclusions about a population?

Machine Learning asks questions such as:

- Can we predict house prices?
- Can we classify an image?
- Can we predict whether a customer will churn?
- Can we recognize spam emails?

Statistics is often used to understand and analyze data, while Machine Learning uses data to build models that can generalize to new observations.

The two fields overlap heavily.

```python
import numpy as np

area = np.array([1000, 1200, 1500, 1800, 2000])
price = np.array([50, 60, 75, 90, 100])

correlation = np.corrcoef(area, price)[0, 1]

print("Correlation between area and price:", correlation)
```

# 4. Descriptive Statistics

Descriptive statistics describes or summarizes the data that we already have.

It does not try to make predictions about a larger population.

Common descriptive statistics include:

- Mean
- Median
- Mode
- Minimum
- Maximum
- Range
- Variance
- Standard deviation
- Percentiles

For example, if we have the marks of 100 students, descriptive statistics can summarize those 100 marks.

```python
import numpy as np

marks = np.array([60, 70, 70, 80, 90])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard Deviation:", np.std(marks))
```

# 5. Inferential Statistics

Inferential statistics uses information from a sample to make conclusions or estimates about a larger population.

For example, suppose a university has 10,000 students.

It may be difficult to ask every student about their satisfaction.

Instead, we can select 500 students and collect their responses.

We can use the sample results to estimate the satisfaction level of the entire student population.

Inferential statistics commonly involves:

- Sampling
- Estimation
- Confidence intervals
- Hypothesis testing
- Statistical tests

```python
import numpy as np

sample = np.array([1, 1, 0, 1, 0, 1, 1, 0, 1, 1])

satisfaction_rate = np.mean(sample) * 100

print("Sample Satisfaction Rate:", satisfaction_rate, "%")
```

# 6. Population

A population is the complete group of individuals or objects that we are interested in studying.

Examples:

- All customers of a company
- All students in a university
- All houses in a city
- All transactions made by a bank

The population can be very large.

When studying a population, collecting information from every member may be expensive or time-consuming.

```python
import numpy as np

population = np.array([100, 200, 300, 400, 500])

population_mean = np.mean(population)

print("Population:", population)
print("Population Mean:", population_mean)
```

# 7. Sample

A sample is a smaller subset selected from a population.

Instead of studying every member of a population, we select some members and study them.

For example:

Population = 100,000 customers

Sample = 1,000 customers

The sample should ideally represent the population reasonably well.

```python
import numpy as np

population = np.arange(1, 101)

sample = np.random.choice(population, size=10, replace=False)

print("Population:", population)
print("Sample:", sample)
```

# 8. Census

A census means collecting information from every member of the population.

For example, if a company has 1,000 employees and collects salary information from all 1,000 employees, it is conducting a census.

Census provides information about the entire population, but it can require more:

- Time
- Money
- Resources

```python
import numpy as np

population = np.array([25000, 30000, 35000, 40000, 45000])

census_mean = np.mean(population)

print("All Employee Salaries:", population)
print("Census Mean Salary:", census_mean)
```

# 9. Sampling

Sampling is the process of selecting a subset of observations from a population.

Sampling is useful when studying the entire population is difficult, expensive, or time-consuming.

Examples of sampling methods include:

- Random sampling
- Systematic sampling
- Stratified sampling
- Cluster sampling

A good sampling method helps produce a sample that represents the population.

```python
import numpy as np

population = np.arange(1, 10001)

sample = np.random.choice(
    population,
    size=500,
    replace=False
)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("First 10 Sample Values:", sample[:10])
```

# 11. Features

A feature is an input variable used by a machine-learning model to learn patterns and make predictions.

For example, if we want to predict house price, possible features are:

- Area
- Number of bedrooms
- Location
- Age of house

The target is the value we want to predict.

Example:

Features → Area, Bedrooms, Age
Target → Price

```python
import pandas as pd

data = pd.DataFrame({
    "Area": [1000, 1500, 2000],
    "Bedrooms": [2, 3, 4],
    "Age": [10, 5, 3],
    "Price": [50, 75, 100]
})

X = data[["Area", "Bedrooms", "Age"]]
y = data["Price"]

print("Features:")
print(X)

print("\nTarget:")
print(y)
```

# 12. Observation

An observation is one individual record or data point in a dataset.

Usually, a row represents one observation.

For example:

| Age | Salary | Experience |
|---:|---:|---:|
| 25 | 30000 | 2 |
| 30 | 50000 | 5 |
| 35 | 70000 | 8 |

Each row is one observation.

Therefore:

Number of observations = Number of rows

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [25, 30, 35],
    "Salary": [30000, 50000, 70000],
    "Experience": [2, 5, 8]
})

print("Number of observations:", len(data))
print("\nFirst observation:")
print(data.iloc[0])
```

# 13. Dataset

A dataset is a structured collection of data containing observations and variables.

In a tabular dataset:

- Rows usually represent observations.
- Columns usually represent variables/features.

Example:

| Age | Salary | Experience |
|---:|---:|---:|
| 25 | 30000 | 2 |
| 30 | 50000 | 5 |
| 35 | 70000 | 8 |

This dataset contains:

3 observations

and

3 variables.

```python
import pandas as pd

dataset = pd.DataFrame({
    "Age": [25, 30, 35],
    "Salary": [30000, 50000, 70000],
    "Experience": [2, 5, 8]
})

print(dataset)

print("\nRows:", dataset.shape[0])
print("Columns:", dataset.shape[1])
```

# 16. Continuous Data

Continuous data can take any value within a range.

It is usually obtained through measurement.

Examples:

- Height
- Weight
- Temperature
- Distance
- Time

For example, height can be:

170 cm
170.2 cm
170.25 cm
170.257 cm

There can theoretically be many possible values between two points.

```python
import numpy as np

temperature = np.array([
    25.1,
    25.5,
    26.2,
    27.8,
    28.4
])

print("Average Temperature:", np.mean(temperature))
print("Minimum Temperature:", np.min(temperature))
print("Maximum Temperature:", np.max(temperature))
```

Discrete data consists of separate, countable values.

It usually represents something that can be counted.

Examples:

- Number of children
- Number of products
- Number of employees
- Number of orders
- Number of website visits

For example, the number of products can be:

1
2
3
4

But normally we would not say:

2.57 products.

```python
import numpy as np

orders = np.array([1, 2, 3, 4, 5])

print("Average Orders:", np.mean(orders))
print("Total Orders:", np.sum(orders))
```

# 18. Structured Data

Structured data is data organized in a predefined format, usually rows and columns.

A relational database table is a common example.

Example:

| CustomerID | Age | Salary |
|---|---:|---:|
| C001 | 25 | 30000 |
| C002 | 30 | 50000 |
| C003 | 35 | 70000 |

The structure makes it easy for computers to store, query, analyze, and process the data.

Common examples include:

- SQL tables
- Excel spreadsheets
- CSV files
- Relational databases

```python
import pandas as pd

data = pd.DataFrame({
    "CustomerID": ["C001", "C002", "C003"],
    "Age": [25, 30, 35],
    "Salary": [30000, 50000, 70000]
})

print(data)
```
